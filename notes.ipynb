##  All about Models in Django
- model is the class which defined inside model.py of django-app
- it inherits the django-models class
- this class represent the table in RDBMS
- the attribute(variable) represent the column of table

```python
from django.db import models
from django.contrib.auth.models import User
gender_choice = (('M','Male'),
                 ('F','Female'),
                 ('O','Other'),)
class Customer(models.Model):                                                   # table
    user  = models.OneToOneField(User,on_delete=models.CASCADE)                 # column 
    gender = models.CharField(choices=gender_choice,max_length=1,default=None)
    D_O_B = models.DateField(default=None )
    def __str__(self) -> str:
        return self.user.username
# https://docs.djangoproject.com/en/6.0/ref/models/options/ 
```

### fileds in Models
#### Text Fields
- CharField: For short-to-medium strings. Requires a max_length.
- TextField: For large amounts of text (e.g., blog posts).
- EmailField: A CharField that validates that the input is a valid email address.
- SlugField: A short label containing only letters, numbers, underscores, or hyphens, often used in URLs.
#### Numeric Fields
- IntegerField: Stores standard integers (usually -2,147,483,648 to 2,147,483,647).
- FloatField: For floating-point numbers (Python float).
- DecimalField: For fixed-precision decimal numbers, ideal for financial data. Requires max_digits and decimal_places.
#### Date and Time Fields
- DateField: Stores only a date (Python datetime.date).
- DateTimeField: Stores both date and time (Python datetime.datetime).
- TimeField: Stores only time (Python datetime.time).
#### Relationship Fields
- ForeignKey: Creates a many-to-one relationship.
- ManyToManyField: Creates a many-to-many relationship.
- OneToOneField: Creates a one-to-one relationship.
#### Other Specialized Fields
- BooleanField: Stores True or False values.
- FileField / ImageField: For file and image uploads (ImageField requires the Pillow library).
- JSONField: (For supported databases) Stores JSON-encoded data.
- UUIDField: For storing Universally Unique Identifiers.
